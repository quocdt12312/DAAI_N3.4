```python
import pandas as pd

# 1. Đọc dữ liệu từ file products.csv
df_products = pd.read_csv('products.csv')

# Kiểm tra nhanh thông tin
print(f"Tổng số sản phẩm ban đầu: {len(df_products)}")
print(f"Số lượng dòng trùng lặp: {df_products.duplicated().sum()}")
```

```python
# 1. Đổi tên cột cho khớp chính xác với Schema yêu cầu (Viết hoa chữ cái đầu theo chuẩn của bạn)
df_products.rename(columns={
    'product_id': 'Product_ID',
    'product_name': 'Product_Name',
    'category': 'Category',
    'price': 'Price'
}, inplace=True)

# 2. Xử lý logic tạo cột Description (Mô tả sản phẩm)
# Kết hợp Tên, Phân loại, Màu sắc và Kích cỡ để tạo mô tả. 
# Ví dụ: "SaigonFlex UC-01 - Streetwear (Màu: green, Size: S)"
df_products['Description'] = (
    df_products['Product_Name'] + ' - ' + 
    df_products['Category'] + 
    ' (Màu: ' + df_products['color'] + ', Size: ' + df_products['size'] + ')'
)

# 3. Làm tròn giá tiền (Price) và giá vốn (cogs) về 2 chữ số thập phân cho gọn gàng
df_products['Price'] = df_products['Price'].round(2)
df_products['cogs'] = df_products['cogs'].round(2)

# 4. Sắp xếp lại thứ tự cột chuẩn xác theo yêu cầu cấu trúc
final_columns = [
    'Product_ID', 'Product_Name', 'Category', 'Price', 
    'Description', 'segment', 'size', 'color', 'cogs'
]
df_products_final = df_products[final_columns]

print("\nDữ liệu mẫu sau khi hoàn thiện Schema:")
display(df_products_final.head())   
```

```python
# 1. Kiểm tra Dữ liệu Missing/Null
print("Kiểm tra Null ở từng cột:")
print(df_products_final.isnull().sum())

# 2. Kiểm tra tính Unique (Duy nhất) của Product_ID
duplicates = df_products_final.duplicated(subset=['Product_ID']).sum()
print(f"\nSố lượng Product_ID bị trùng lặp: {duplicates}")

if duplicates > 0:
    df_products_final = df_products_final.drop_duplicates(subset=['Product_ID'])
    print("-> Đã xóa các sản phẩm trùng lặp.")

# 3. Xuất ra file chuẩn hóa
output_file = 'products_silver.csv'
df_products_final.to_csv(output_file, index=False, encoding='utf-8')

print(f"\n✅ Đã xử lý xong và lưu {len(df_products_final)} sản phẩm vào file {output_file}!")
```
